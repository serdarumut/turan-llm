```python
import torch
import torch.nn as nn

torch.manual_seed(42)

vocab_size = 10
embedding_dim = 4

embedding = nn.Embedding(vocab_size, embedding_dim)

print(embedding.weight.shape)
print(embedding.weight)
```

```python
ids = torch.tensor([3, 7, 3])

vectors = embedding(ids)

print(vectors.shape)
print(vectors)
```

```python
batch = torch.tensor([[3, 7, 3],
                      [1, 2, 5]])

out = embedding(batch)

print(batch.shape)
print(out.shape)
```

```python
context_length = 3
pos_embedding = nn.Embedding(context_length, embedding_dim)

positions = torch.arange(3)
print(positions)

x = embedding(ids) + pos_embedding(positions)
print(x)
```

```python
import torch.nn.functional as F

kedi = embedding.weight[3]
kopek = embedding.weight[7]

print("kedi - köpek:", F.cosine_similarity(kedi, kopek, dim=0))
print("kedi - kedi :", F.cosine_similarity(kedi, kedi, dim=0))
```
